```python
#Importar librería para abrir y leer archivos PDF
import pdfplumber
import pandas as pd
#Modulo re, permite trabajar con expresiones regulares
#para buscar patrones dentro de texto
```

```python
#Crear una cadena de texto vacía
texto = " "

#Abrir el PDF e iterar por sus páginas para extraer el texto:
with pdfplumber.open("/content/datos_estudiantes.pdf") as pdf:
  for pagina in pdf.pages:
    texto += pagina.extract_text() + "\n"
    print(texto)
```

```python
import re

patron_demografico = re.compile(
    #^ indica que la búsqueda debe de comenzar al inicio de una línea
    #(\d{3}) busca exactamente 3 dígitos: corresponde al ID
    #\s+ representa uno o más espacios
    #(.+?) captura el nombre del estudiante
    #(\d+) captura la edad
    #(\d+) captura el semestre
  r"^(\d{3})\s+(.+?)\s+(\d+)\s+(\d+)"

#Busca una de las tres carreras posibles
  r"(Ingeniería en Software|Ciencia de Datos|Sistemas Computacionales)"

  #Busca una de las dos modalidades posibles
  r"(Presencial|Híbrida)$",

  re.MULTILINE
)
```

```python
#Busca dentro de "texto" todas las líneas que cumplen con el patrón que acabamos de definir
datos_demograficos = patron_demografico.findall(texto)

datos_demograficos
```

```python
#Convertir los datos demograficos en un DataFra,e de pandas
df_demograficos = pd.DataFrame(
    datos_demograficos,
    columns=["ID", "Nombre", "Edad", "Semestre", "Carrera", "Modalidad"]
)
df_demograficos.head()
```
